# Cryptography — 20 Illustrated ATLAS Experiments

The original security experiments run real repository code. Each experiment produces **two distinct figures**: an actual execution/status chart, and a **conceptual model that is NOT experimentally measured**. Both are labeled. Logs, CSV, JSON and PNG files can be downloaded. Do not run the toy cryptographic examples to protect real secrets.

Guide: https://github.com/FUllSCIPTS/ATLAS/blob/main/docs/CRYPTOGRAPHY.md


```python
import subprocess, pathlib, sys
root=pathlib.Path("ATLAS")
if not (root/"security_lab"/"visual_lab.py").exists():
    subprocess.run(["git","clone","--depth","1","https://github.com/FUllSCIPTS/ATLAS.git","ATLAS"],check=True)
import importlib.util
spec=importlib.util.spec_from_file_location("atlas_visual",root/"security_lab"/"visual_lab.py")
visual=importlib.util.module_from_spec(spec)
spec.loader.exec_module(visual)
from IPython.display import display, Image
from pathlib import Path
OUT=Path("atlas_visual_results")
DOMAIN="crypto"
print("Ready:", DOMAIN, "20 source experiments")
```

## Instructions

For each numbered section, first predict the result using the theory guide. Run the cell. **Observed** plots show execution time and process exit status; the separate **concept** plot is an explanatory model, not results of the source experiment. Read the per-experiment log before drawing conclusions. A run marked FAIL requires inspecting its log; the notebook does not conceal failures.


```python
def view_experiment(number):
    result=visual.run_one(DOMAIN, number, OUT, timeout=300)
    print("Source runner:",visual.SCRIPTS[DOMAIN].name,"| exit:",result["exit_code"],"| elapsed:",result["elapsed_seconds"],"s")
    folder=OUT/DOMAIN
    display(Image(filename=str(folder/f"{number:02}_observed.png")))
    display(Image(filename=str(folder/f"{number:02}_concept.png")))
    log=(folder/f"{number:02}_run.log").read_text(encoding="utf8")
    print("LOG (first 2800 chars):\n",log[:2800])
    if not result["success"]: print("CHECK FAILURE: see full .log in exported archive")
    return result
RESULTS=[]
```

## Results, audit and export

Failures are retained as findings, not silently discarded. Summarize which tests passed, chart their execution times, and export the raw logs plus illustrative/observed graphics. Do not conflate toy conceptual curves with empirical measurements.


```python
import csv, shutil
OUT.mkdir(exist_ok=True)
with (OUT/f"{DOMAIN}_summary.csv").open("w",newline="") as f:
    writer=csv.DictWriter(f,fieldnames=list(RESULTS[0]))
    writer.writeheader();writer.writerows(RESULTS)
print("Passed",sum(r["success"] for r in RESULTS),"of",len(RESULTS))
print("Conceptual charts are NOT measured results.")
shutil.make_archive(f"{DOMAIN}_20_labs","zip",OUT)
print("Download:",f"{DOMAIN}_20_labs.zip")
# In Colab, optionally use files.download(f"{DOMAIN}_20_labs.zip")
```
