Companion to the [Python course](https://github.com/FUllSCIPTS/ATLAS/blob/main/docs/PYTHON_PROGRAMMING.md). **Runtime → Run all** takes under a minute. Nothing here needs an account or real personal data.

```python
# Setup: fetch the repository (for the ATLAS Go code used in some examples) and load the lab module.
import importlib.util, pathlib, subprocess, sys
from pathlib import Path
from IPython.display import Image, display
if Path('python_lab/experiments.py').exists():
    ROOT = Path('.')
elif Path('experiments.py').exists() and Path('../android-app').exists():
    ROOT = Path('..')
else:
    if not Path('ATLAS').exists():
        subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/FUllSCIPTS/ATLAS.git'], check=True)
    ROOT = Path('ATLAS')
sys.path.insert(0, str(ROOT / 'android-app'))
spec = importlib.util.spec_from_file_location('plab', ROOT / 'python_lab' / 'experiments.py')
plab = importlib.util.module_from_spec(spec); sys.modules['plab'] = plab; spec.loader.exec_module(plab)
OUT = Path('python_results'); OUT.mkdir(exist_ok=True)

def lab(number):
    res = plab.experiment(number)
    png = plab.save_experiment(res, OUT)
    print(f"P{number:02} {res['title']}")
    print('  ', res['xlabel'], '|', ' | '.join(res['labels']))
    for row in res['rows']:
        print('  ', ' | '.join(f'{v:.4g}' if isinstance(v, float) else str(v) for v in row))
    if res['notes']:
        print('Note:', res['notes'])
    display(Image(filename=str(png)))
```

## P15 — Generators

**Try it — modules and packages:** this cell writes a `physics` package (`__init__.py`, `module1.py`, `module2.py`, `particle.py`) and a `main.py` next to it, runs `main.py`, then tries the four ways to import `module5.sum`. Edit a file's text and re-run.

```python
import subprocess, sys, textwrap
from pathlib import Path

demo = Path('modules_demo')
files = {
    'physics/__init__.py': """
        from .module1 import distance, speed
        from .module2 import kinetic_energy
    """,
    'physics/module1.py': """
        def speed(distance_m, time_s):
            return distance_m / time_s

        def distance(speed_ms, time_s):
            return speed_ms * time_s
    """,
    'physics/module2.py': """
        def kinetic_energy(mass_kg, speed_ms):
            return 0.5 * mass_kg * speed_ms ** 2
    """,
    'physics/particle.py': """
        from .module1 import speed
        from .module2 import kinetic_energy

        class Particle:
            def __init__(self, mass_kg):
                self.mass_kg = mass_kg

            def energy_after(self, distance_m, time_s):
                return kinetic_energy(self.mass_kg, speed(distance_m, time_s))
    """,
    'main.py': """
        import physics                          # the folder name
        from physics.particle import Particle   # one class from one module

        print(round(physics.speed(100, 9.58), 2), 'm/s', round(Particle(94).energy_after(100, 9.58)), 'J')
    """,
    'module5.py': """
        def sum(a, b):
            return a + b
    """,
}
for name, body in files.items():
    path = demo / name
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(textwrap.dedent(body).lstrip())
print(subprocess.run([sys.executable, 'main.py'], cwd=demo, capture_output=True, text=True).stdout)

sys.path.insert(0, str(demo.resolve()))
import module5                      # module5.sum(2, 3); the built-in sum still works
from module5 import sum as add      # rename on import: no clash with the built-in sum
import module5 as m5
print(module5.sum(2, 3), add(2, 3), m5.sum(2, 3), sum([1, 2, 3]))
# Try: write  from module5 import sum  and then  sum([1, 2, 3])  — what happens, and why?
```
