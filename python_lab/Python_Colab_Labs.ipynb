# Python with ATLAS Go — 20 illustrated labs, a testing workshop and 12 app-integration labs

Companion to the [Python course](https://github.com/FUllSCIPTS/ATLAS/blob/main/docs/PYTHON_PROGRAMMING.md). **Runtime → Run all** takes about a minute. Nothing here needs an account or real personal data.

```python
# Setup: fetch the repository (for the ATLAS Go code used in some examples) and load the lab module.
import importlib.util, pathlib, subprocess, sys
from pathlib import Path
from IPython.display import Image, display
if Path('python_lab/experiments.py').exists():
    ROOT = Path('.')
elif Path('experiments.py').exists() and Path('../android-app').exists():
    ROOT = Path('..')
else:
    if not Path('ATLAS').exists():
        subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/FUllSCIPTS/ATLAS.git'], check=True)
    ROOT = Path('ATLAS')
sys.path.insert(0, str(ROOT / 'android-app'))
spec = importlib.util.spec_from_file_location('plab', ROOT / 'python_lab' / 'experiments.py')
plab = importlib.util.module_from_spec(spec); sys.modules['plab'] = plab; spec.loader.exec_module(plab)
OUT = Path('python_results'); OUT.mkdir(exist_ok=True)

def lab(number):
    res = plab.experiment(number)
    png = plab.save_experiment(res, OUT)
    print(f"P{number:02} {res['title']}")
    print('  ', res['xlabel'], '|', ' | '.join(res['labels']))
    for row in res['rows']:
        print('  ', ' | '.join(f'{v:.4g}' if isinstance(v, float) else str(v) for v in row))
    if res['notes']:
        print('Note:', res['notes'])
    display(Image(filename=str(png)))
```

**Try it — modules and packages:** this cell writes a `physics` package (`__init__.py`, `module1.py`, `module2.py`, `particle.py`) and a `main.py` next to it, runs `main.py`, then tries the four ways to import `module5.sum`. Edit a file's text and re-run.

```python
import subprocess, sys, textwrap
from pathlib import Path

demo = Path('modules_demo')
files = {
    'physics/__init__.py': """
        from .module1 import distance, speed
        from .module2 import kinetic_energy
    """,
    'physics/module1.py': """
        def speed(distance_m, time_s):
            return distance_m / time_s

        def distance(speed_ms, time_s):
            return speed_ms * time_s
    """,
    'physics/module2.py': """
        def kinetic_energy(mass_kg, speed_ms):
            return 0.5 * mass_kg * speed_ms ** 2
    """,
    'physics/particle.py': """
        from .module1 import speed
        from .module2 import kinetic_energy

        class Particle:
            def __init__(self, mass_kg):
                self.mass_kg = mass_kg

            def energy_after(self, distance_m, time_s):
                return kinetic_energy(self.mass_kg, speed(distance_m, time_s))
    """,
    'main.py': """
        import physics                          # the folder name
        from physics.particle import Particle   # one class from one module

        print(round(physics.speed(100, 9.58), 2), 'm/s', round(Particle(94).energy_after(100, 9.58)), 'J')
    """,
    'module5.py': """
        def sum(a, b):
            return a + b
    """,
}
for name, body in files.items():
    path = demo / name
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(textwrap.dedent(body).lstrip())
print(subprocess.run([sys.executable, 'main.py'], cwd=demo, capture_output=True, text=True).stdout)

sys.path.insert(0, str(demo.resolve()))
import module5                      # module5.sum(2, 3); the built-in sum still works
from module5 import sum as add      # rename on import: no clash with the built-in sum
import module5 as m5
print(module5.sum(2, 3), add(2, 3), m5.sum(2, 3), sum([1, 2, 3]))
# Try: write  from module5 import sum  and then  sum([1, 2, 3])  — what happens, and why?
```

# Part III — Building real apps: integration flows (A01–A12)

One button click in a real app crosses five layers: **button → event handler → HTTP client → API route → service → SQLite**, and the answer climbs back to the screen. Each lab asserts what must be true and charts what it observed. [Read Part III of the guide](https://github.com/FUllSCIPTS/ATLAS/blob/main/docs/PYTHON_PROGRAMMING.md#part-iii--building-real-apps-integration-flows).

| A05 | double-click protection and idempotency keys |

```python
# Setup for Part III: load the integration-lab module (ROOT comes from the first setup cell).
import importlib.util, sys
spec = importlib.util.spec_from_file_location('app_integration', ROOT / 'python_lab' / 'app_integration.py')
ai = importlib.util.module_from_spec(spec); sys.modules['app_integration'] = ai; spec.loader.exec_module(ai)
from app_integration import App, RideScreen
APP_OUT = Path('app_results'); APP_OUT.mkdir(exist_ok=True)

def app_lab(number):
    res = ai.experiment(number)
    png = ai.save_experiment(res, APP_OUT)
    print(f"A{number:02} {res['title']}")
    print('  ', res['xlabel'], '|', ' | '.join(res['labels']))
    for row in res['rows']:
        print('  ', ' | '.join(f'{v:.4g}' if isinstance(v, float) else str(v) for v in row))
    for key, lines in res['extra'].items():
        print(f'{key}:'); print('\n'.join('   ' + line for line in lines))
    if res['notes']:
        print('Note:', res['notes'])
    display(Image(filename=str(png)))
```

## A04 — Search-as-you-type and debouncing

A request per key press multiplies traffic and SQL work; debouncing waits for a pause. User text is passed to `LIKE ?` as a parameter with `%` and `_` escaped. [Read the lesson](https://github.com/FUllSCIPTS/ATLAS/blob/main/docs/PYTHON_PROGRAMMING.md#a04--search-as-you-type-and-debouncing).

**Try it:**

```python
events = ai.keystrokes('airport', pauses={3: 500})
print(events)
for delay in (100, 300, 800):
    print(delay, 'ms ->', ai.debounced(events, delay))
```

## A05 — Double submit

A second click can arrive while the first request is in flight. Disable the button *and* send an `Idempotency-Key`. [Read the lesson](https://github.com/FUllSCIPTS/ATLAS/blob/main/docs/PYTHON_PROGRAMMING.md#a05--double-submit).

**Try it:**

```python
for disable, key in [(False, False), (False, True), (True, False)]:
    with App(drivers=1, disable_while_sending=disable, send_request_key=key) as app:
        app.server.faults['/api/rides'] = ai.deque(['delay:0.1', 'delay:0.1'])
        app.screen.form.update(passenger='Aoife', pickup='UCC', destination='Douglas', km='4')
        app.screen.book_button.click(); app.screen.book_button.click(); app.screen.wait()
        print(f'disable={disable!s:5} key={key!s:5}: {app.client.requests} requests -> {app.db.count("rides")} rides')
```

## A06 — Load more: offset vs keyset

`OFFSET` counts positions, so a row inserted between clicks makes a row appear twice; `WHERE id < last_seen_id` cannot repeat. [Read the lesson](https://github.com/FUllSCIPTS/ATLAS/blob/main/docs/PYTHON_PROGRAMMING.md#a06--load-more-offset-vs-keyset).

**Try it:**

```python
with App(drivers=1, rides=25, page_size=10, pagination='offset') as app:
    app.screen.load_button.click()
    app.client.post('/api/rides', {'passenger': 'New', 'pickup': 'UCC', 'destination': 'Douglas', 'distance_km': 3})
    app.screen.more_button.click()
    ids = [r['id'] for r in app.screen.table]
    print(len(ids), 'rows,', len(ids) - len(set(ids)), 'duplicate(s)')
# Try: pagination='keyset'
```

## A07 — Concurrent edits

Without a version check the second save erases the first (a lost update). With `WHERE id = ? AND version = ?` the stale save gets 409 Conflict. [Read the lesson](https://github.com/FUllSCIPTS/ATLAS/blob/main/docs/PYTHON_PROGRAMMING.md#a07--concurrent-edits).

**Try it:**

```python
with App(drivers=1, rides=1) as app:
    bob = RideScreen(app.client, app.trace)
    app.screen.open_button.click(1); bob.open_button.click(1)
    app.screen.editing['passenger'] = 'Alice'; print('Alice:', app.screen.save_button.click().status)
    bob.editing['passenger'] = 'Bob';          print('Bob:  ', bob.save_button.click().status, '-', bob.banner)
    bob.close()
# Try: RideScreen(..., use_versions=False) for both screens
```

## A08 — Transactions

*Assign* writes three rows (ride, driver, audit). A crash after the first write leaves the data inconsistent unless the three writes share one transaction. [Read the lesson](https://github.com/FUllSCIPTS/ATLAS/blob/main/docs/PYTHON_PROGRAMMING.md#a08--transactions).

**Try it:**

```python
for atomic in (False, True):
    with App(drivers=2, rides=2) as app:
        app.service.atomic_assign = atomic
        app.service.crash_assign_ids = {1}
        resp = app.screen.assign_button.click(1, 1)
        print(f'atomic={atomic}: HTTP {resp.status}, ride status={app.service.get_ride(1)["status"]}, '
              f'busy drivers={app.db.count("drivers", "busy = 1")}, banner={app.screen.banner!r}')
```

A cache saves queries but can show stale data; clear it on every write that changes what it summarises. [Read the lesson](https://github.com/FUllSCIPTS/ATLAS/blob/main/docs/PYTHON_PROGRAMMING.md#a10--caching-and-invalidation).

## A11 — Errors, timeouts and retries

Retry only transient failures (503, timeouts, refused connections) with back-off; never 4xx or 500. Every failure becomes a plain sentence. [Read the lesson](https://github.com/FUllSCIPTS/ATLAS/blob/main/docs/PYTHON_PROGRAMMING.md#a11--errors,-timeouts-and-retries).

**Try it:**

```python
with App(drivers=1) as app:
    for fault in ([503, 503], [500], ['delay:0.6']):
        app.server.faults['/api/stats'] = ai.deque(fault)
        out = app.screen.refresh_button.click()
        print(f'{str(fault):14} -> HTTP {out.status} after {out.attempts} attempt(s); banner: {app.screen.banner or "data shown"}')
```

**Lab:** predict the chart, then run.
